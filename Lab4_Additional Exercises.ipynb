```python
# --- EXERCISE 2: JACCARD SIMILARITY ENGINE ---

# Required Inputs (Carried over)
FAVORITE_ARTIST = "BLACKPINK"
CONTROL_NUM = 3

# 1. Initialize Sets
festival = {
    "Ben&Ben", "SB19", "Bini", "Eraserheads", 
    FAVORITE_ARTIST, "Zild", f"Indie Artist {CONTROL_NUM}"
}

user_a = {"Ben&Ben", "Bini", "Maki", "Dionela", FAVORITE_ARTIST}
user_b = {
    "SB19", "Eraserheads", "Zild", 
    f"Indie Artist {CONTROL_NUM}", "Parokya ni Edgar"
}

# 2. Data Expansion: Add exactly CONTROL_NUM (3) new artists
new_artists = {"Lola Amour", "Dilaw", "Juan Karlos"}
festival.update(new_artists)

# 3. Algorithm: Jaccard Similarity J(A, B) = |A ∩ B| / |A ∪ B|

# User A Calculations
intersection_a = user_a.intersection(festival)
union_a = user_a.union(festival)
jaccard_a = (len(intersection_a) / len(union_a)) * 100

# User B Calculations
intersection_b = user_b.intersection(festival)
union_b = user_b.union(festival)
jaccard_b = (len(intersection_b) / len(union_b)) * 100

# 4. Set Difference: User A's Dealbreakers (In User A but NOT in Festival)
dealbreakers_a = user_a.difference(festival)

print("="*45)
print("       EXERCISE 2 ASSESSMENT DATA")
print("="*45)
print(f"Total Artists in Expanded Festival   : {len(festival)}")
print(f"Length of Intersection (User A & Fest): {len(intersection_a)}")
print(f"Length of Union (User A & Festival)  : {len(union_a)}")
print(f"User A Jaccard Similarity (%)        : {jaccard_a:.2f}%")
print(f"User B Jaccard Similarity (%)        : {jaccard_b:.2f}%")
print(f"User A Dealbreaker Artists           : {dealbreakers_a}")
print("="*45)
```

```python
from collections import deque, Counter

# 1. Initialize deque with maxlen = 5
rolling_buffer = deque(maxlen=5)

# 2. Simulate burst of incoming alerts
burst = [ 
    (CONTROL_NUM, "WEATHER"),           # 3
    (CONTROL_NUM + 1, "TRAFFIC"),       # 4
    (CONTROL_NUM + 2, "WEATHER"),       # 5
    (CONTROL_NUM + 3, "WEATHER"),       # 6
    (CONTROL_NUM + 4, LOCAL_HAZARD),    # 7
    (CONTROL_NUM + 5, "WEATHER"),       # 8
    (CONTROL_NUM + 6, "TRAFFIC")        # 9
] 

# 3. Data Expansion: Add exactly (CONTROL_NUM + 2) = 5 custom alerts
# Starting sequentially from Timestamp 10
last_ts = burst[-1][0]
for i in range(1, CONTROL_NUM + 3):
    # Adding a variety of categories to test the spam rule
    category = "TRAFFIC" if i % 2 == 0 else "FIRE"
    burst.append((last_ts + i, category))

# 4. Algorithm (The Sliding Window)
rejections = {} # To track specific assessment questions

for timestamp, category in burst:
    # Extract categories currently in the deque
    current_categories = [item[1] for item in rolling_buffer]
    category_counts = Counter(current_categories)
    
    # Spam Rule: Reject if category exists 2 or more times
    if category_counts[category] >= 2:
        rejections[timestamp] = "N"
        continue 
    else:
        rolling_buffer.append((timestamp, category))
        rejections[timestamp] = "Y"

# 5. Final Calculations
final_ts_sum = sum(alert[0] for alert in rolling_buffer)

print("="*50)
print("       EXERCISE 3 ASSESSMENT DATA")
print("="*50)
print(f"LOCAL_HAZARD Used                     : {LOCAL_HAZARD}")
print(f"Total Burst Size (Expanded)           : {len(burst)}")
print(f"WEATHER at {CONTROL_NUM + 3} append?           : {rejections[CONTROL_NUM + 3]}")
print(f"LOCAL_HAZARD append?                  : {rejections[CONTROL_NUM + 4]}")
print(f"Final Deque Length                    : {len(rolling_buffer)}")
print(f"Final Sum of Timestamps in Deque      : {final_ts_sum}")
print(f"Exact Final Deque Output              : {list(rolling_buffer)}")
print("="*50)
```
